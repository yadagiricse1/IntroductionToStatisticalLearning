```python
auto = pd.read_csv("https://www.statlearning.com/s/Auto.csv")
auto
```

```python
auto.describe()
```

```python
import numpy as np
auto.loc[auto['horsepower'] == '?','horsepower'] = np.nan
auto['horsepower'] = auto['horsepower'].astype(float)
auto.dropna(inplace=True)
auto.reset_index(inplace=True,drop=True)
```

```python
quant = ['mpg','displacement','horsepower','weight','acceleration']
pd.DataFrame([auto[quant].min().values,auto[quant].max().values],columns=quant, index=['min','max'])
```

```python
pd.DataFrame([auto[quant].mean().values,auto[quant].std().values],columns=quant, index=['mean','std'])
```

```python
auto[quant].drop(index=range(10,85)).describe()
```

```python
pd.plotting.scatter_matrix(auto);
```

```python
pd.plotting.boxplot(auto, column='mpg', by='year')
```
